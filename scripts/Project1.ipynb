```python
# Read the sequence and concatenate the sequence lines
genome = ""
with open ("MT-CO1.fasta") as fh:
    for line in fh:
        if not line.startswith(">"):
            genome = genome + line.strip()
            
#Extract cds 
cds = genome[5903:7445]
 
#Find and print cds lenght
print("CDS length: ", len(cds))

#divisibility by 3
print("Divisible by three: ", len(cds) % 3 == 0)
```

```python
#build genetic code to translate entire coding sequence to protein
#In mitochondria, ATA = M, TGA = W, AGA= stop, AGG= stop
#Here we used * for stop codon 
mito_code = { "CTT" : "L" , "CTC" : "L" , "CTA" : "L" , "CTG" : "L",
              "TTT" : "F" , "TTC" : "F" , "TTA" : "L" , "TTG" : "L",
              "ATT" : "I" , "ATC" : "I" , "ATA" : "M" , "ATG" : "M",
              "GTT" : "V" , "GTC" : "V" , "GTA" : "V" , "GTG" : "V",
              "CCT" : "P" , "CCC" : "P" , "CCA" : "P" , "CCG" : "P",
              "TCT" : "S" , "TCC" : "S" , "TCA" : "S" , "TCG" : "S",
              "ACT" : "T" , "ACC" : "T" , "ACA" : "T" , "ACG" : "T",
              "GCT" : "A" , "GCC" : "A" , "GCA" : "A" , "GCG" : "A",
              "CAT" : "H" , "CAC" : "H" , "CAA" : "Q" , "CAG" : "Q",
              "TAT" : "Y" , "TAC" : "Y" , "TAA" : "*" , "TAG" : "*",
              "AAT" : "N" , "AAC" : "N" , "AAA" : "K" , "AAG" : "K",
              "GAT" : "D" , "GAC" : "D" , "GAA" : "E" , "GAG" : "E",
              "CGT" : "R" , "CGC" : "R" , "CGA" : "R" , "CGG" : "R",
              "TGT" : "C" , "TGC" : "C" , "TGA" : "W" , "TGG" : "W",
              "AGT" : "S" , "AGC" : "S" , "AGA" : "*" , "AGG" : "*", 
              "GGT" : "G" , "GGC" : "G" , "GGA" : "G" , "GGG" : "G",
}  
# Translate the cds
#read three bases at a time
protein = ""
for i in range(0, len(cds),3):
    codon = cds[i:i+3]
    A = mito_code[codon]
    if A == "*":
        break
    protein = protein + A

print("First 30 residues: ", protein[:30])
print("Total lenght: ", len(protein))
```

```python
#read the deposited protein
deposited = ""
with open ("COX1_Protein.fasta") as fh:
    for line in fh:
        if not line.startswith(">"):
            deposited = deposited + line.strip()

#compare the two sequences
if protein == deposited:
    print("Sequences are identical")
else:
    print("Sequences are not identical")

#Find the differences
for i in range(min(len(protein), len(deposited))):
         if protein[i] != deposited[i]:
             print("First difference position: ", i + 1)
             print("My translation sequence:", protein[i])
             print("Deposited protein sequence: ", deposited[i])
             break
            
```

```python

from Bio.Seq import Seq

#translate cds usong Biopython's builtin tranlate function
#table = 2 selects Vertebrate Mitochondrial Code
#cds= True tells Biopython to expect a complete coding seq
biopython_translation = str(Seq(cds). translate(table = 2, cds=True))

#compare the seq
print("Biopython length: ", len(biopython_translation))
print("My translation length: ", len(protein))
print("Biopython agrees: ", biopython_translation == protein)
```
